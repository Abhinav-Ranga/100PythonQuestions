# q 11:  Write a program to find the simple interest when the value of principle,rate of interest and time period is given.

```python
tp = float(input('enter the timePeriod: '))
rp = float(input('enter the rateofinterest: '))
pr =  float(input('enter the principle: '))


res = (pr*rp*tp)/100 

print(res)

```
