```python
def finder(hot , humid):
    if hot >= 30 and humid>=90 :
        print('it is hot and humid')
    elif hot>=30  and humid<90:
        print('this is hot')
    elif hot<30 and humid >=90:
        print('cool and humid')
    elif hot<30 and humid<90:
        print('cool')

hot = int(input('enter the hot: '))        
humid = int(input('enter the humid: '))        

finder(hot,humid)
        
```
