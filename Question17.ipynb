```python
def squareCalculator():
    digit1 = int(input('Enter the digit no 1: '))
    digit2 = int(input('Enter the digit no 2: '))
    digit3 = int(input('Enter the digit no 3: '))

    sum = digit1 **2 + digit2 **2 + digit3 **2
    return print('ur square sum of all digits is : ',sum)
    
squareCalculator()
```
