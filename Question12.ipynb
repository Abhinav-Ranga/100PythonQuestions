```python
import math
def volumefinder(cost_per_litre,radius,height):
    volume =  math.pi*( radius**2)*height
    volume_litres = volume/1000

    total_cost = volume_litres*cost_per_litre

    return total_cost

radius = int(input("enter the radius:"))
height = int(input('enter the height:'))
cost_per = int(input('enter the costperlitre:'))

volumefinder(cost_per,radius,height)
    
    
```
