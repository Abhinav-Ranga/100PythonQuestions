# Q8: Write a program to find the euclidean distance between two coordinates.


```python
import math

x1 = int(input("Enter the x1: "))
y1 = int(input("Enter the y1: "))
x2 = int(input("Enter the x2: "))
y2 = int(input("Enter the y2: "))

o = (x2-x1)**2 + (y2-y1) ** 2
result = math.sqrt(o)
print(result)
```
