# q 10 : Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit

```python


cp = int(input("enter the cost price: "))
sp = int(input("enter the selling price: "))

if sp<cp :
    print("loss: ", cp-sp)
elif sp == cp:
    print("neutral")
else :
    print(" profit: ", sp-cp)
```
