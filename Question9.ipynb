# Q9 Write a program that take a user input of three angles and will find out whether it can form a triangle or not.

```python
a1 = int(input('enter the angle 1: '))
a2 = int(input('enter the angle 2: '))
a3 = int(input('enter the angle 3: '))



if a1 >0 and a2>0 and a3>0  and a1+a2+a3 == 180:
  print('valid')
else:
  print('invalid')
```
